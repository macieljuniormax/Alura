# Testes Replicáveis, Estratificação e Leitura

Este notebook tem como objetivo realizar uma análise de dados de tracking e construir um modelo preditivo para determinar se um usuário comprou ou não um produto com base em diversas variáveis. 

## Conteúdo do Notebook

1. **Importação de Bibliotecas e Leitura de Dados**: Importamos as bibliotecas necessárias e lemos os dados a partir de um arquivo CSV.
2. **Exploração Inicial dos Dados**: Visualizamos as primeiras linhas do dataset e verificamos a estrutura dos dados.
3. **Preparação dos Dados**: Separamos as variáveis independentes (features) e a variável dependente (target).
4. **Divisão dos Dados**: Dividimos os dados em conjuntos de treino e teste para validar o modelo.
5. **Treinamento e Avaliação do Modelo**: Utilizamos um modelo de Support Vector Machine (SVM) para treinar e avaliar a acurácia do modelo.
6. **Utilização de Bibliotecas para Separação de Treino e Teste**: Demonstramos como utilizar a biblioteca `scikit-learn` para realizar a separação dos dados de forma mais eficiente.

Ao final, avaliamos a performance do modelo utilizando a métrica de acurácia.

```python
import pandas as pd

dados = pd.read_csv('tracking.csv', delimiter = ',')
dados.head()
```

```python
y = dados["comprou"]
y.head()
```

```python
x = dados[["inicial",	"palestras",	"contato",	"patrocinio"]]
x.head()
```

```python
dados.shape
```

```python
treino_x = x[:75]
treino_y = y[:75]

teste_x = x[75:]
teste_y = y[75:]

print(f"Treinaremos com {len (treino_x)} elementos")
print(f"Testaremos com {len(teste_x)} elementos")
```

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

modelo = LinearSVC()
modelo.fit(treino_x, treino_y)

previsoes = modelo.predict(teste_x)

acuracia = accuracy_score(teste_y, previsoes) * 100

print(f"A acurácia foi de {acuracia:.2f}%")
```

## Utilizando a biblioteca para separar treino e teste

```python
from sklearn.model_selection import train_test_split

treino_x, teste_x, treino_y, teste_y =  train_test_split(x, y, random_state = 4364, stratify = y)

print(f"Treinaremos com {len (treino_x)} elementos")
print(f"Testaremos com {len(teste_x)} elementos")

modelo = LinearSVC()
modelo.fit(treino_x, treino_y)

previsoes = modelo.predict(teste_x)

acuracia = accuracy_score(teste_y, previsoes) * 100

print(f"A acurácia foi de {acuracia:.2f}%")
```
